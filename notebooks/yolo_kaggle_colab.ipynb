```python
# Tự động tìm thư mục chứa ảnh gốc traffic_train/images trên Kaggle / Colab / Local
search_roots = [
    Path('za_traffic_2020/traffic_train/images'),
    Path('../za_traffic_2020/traffic_train/images'),
    Path('/kaggle/input'),
    Path('/content'),
]

IMAGES_DIR = None
for root in search_roots:
    if not root.exists():
        continue
    if root.name == 'images' and any(root.glob('*.png')):
        IMAGES_DIR = root.resolve()
        break
    for p in root.rglob('traffic_train/images'):
        if p.is_dir():
            IMAGES_DIR = p.resolve()
            break
    if IMAGES_DIR:
        break

assert IMAGES_DIR is not None, (
    'Không tìm thấy thư mục traffic_train/images! '
    'Trên Kaggle hãy nhấn Add Input -> thêm dataset za-traffic-2020.'
)
print('Đã tìm thấy thư mục ảnh gốc tại:', IMAGES_DIR)
print('Số lượng file ảnh:', len(list(IMAGES_DIR.glob('*.png'))))
```

## 5. Tuần 4 – Phần mở rộng riêng của B: Tiling khi huấn luyện + Suy luận SAHI
1. **Suy luận SAHI (Slicing Aided Hyper Inference):** Cắt ảnh 1622×626 thành các mảnh 640×640 chồng lấp 20%, dự đoán trên từng mảnh + toàn ảnh rồi gộp bằng class-wise NMS.
2. **Huấn luyện trên ảnh Tiling (`scripts/tile_yolo_dataset.py`):** Cắt tập train thành các tile 640×640 để mô hình học trực tiếp biển báo ở độ phân giải gốc.

```python
# 5.1. Suy luận SAHI trên mô hình yolov8s_640 để đo mức tăng AP_small không cần train lại
!python scripts/predict_yolo.py --weights runs/detect/yolov8s_640/weights/best.pt \
    --gt data/splits/val.json --images data/yolo/images/val \
    --imgsz 640 --sahi --slice-size 640 --overlap 0.2 \
    --out preds/yolov8s_640_sahi_val.json --eval --name yolov8s_640_sahi
```

```python
# 5.2. Tạo dữ liệu Tiling 640x640 và huấn luyện YOLOv8s trên tập Tiled + suy luận SAHI
!python scripts/tile_yolo_dataset.py --splits data/splits --images "{IMAGES_DIR}" --out data/yolo_tiled

!python scripts/train_yolov8.py --data data/yolo_tiled/data.yaml --model yolov8s.pt \
    --imgsz 640 --epochs 40 --batch 16 --name yolov8s_tiled_640 --exist-ok {WANDB_FLAG}

!python scripts/predict_yolo.py --weights runs/detect/yolov8s_tiled_640/weights/best.pt \
    --gt data/splits/val.json --images data/yolo_tiled/images/val \
    --imgsz 640 --sahi --slice-size 640 --overlap 0.2 \
    --out preds/yolov8s_tiled_sahi_val.json --eval --name yolov8s_tiled_sahi
```

## 6. Tổng hợp bảng kết quả & Đóng gói sản phẩm bàn giao cho A và C
- Xem bảng tổng hợp `reports/eval/results.md`.
- Nén toàn bộ `preds/*.json` (gửi A ở Bàn giao #5), `reports/eval/*`, và trọng số `best.pt` (gửi C ở Bàn giao #8) thành `b_yolo_deliverables.zip` để tải về máy.

```python
# 6.1. Tự động tìm các file best.pt đã train (trong runs/ hoặc /kaggle/input) và xuất đủ bộ preds/*.json
from pathlib import Path
import zipfile

for z in Path('/kaggle/input').rglob('*.zip'):
    with zipfile.ZipFile(z, 'r') as zf:
        zf.extractall('.')

def find_pt(exp_name):
    for root in [Path('.'), Path('/kaggle/input')]:
        if root.exists():
            matches = [p for p in root.rglob('best.pt') if f'/{exp_name}/' in p.as_posix()]
            if matches:
                return str(matches[0].resolve())
    return None

for exp, imgsz, extra in [
    ('yolov8s_640', 640, ''),
    ('yolov8s_640', 640, '--sahi --slice-size 640 --overlap 0.2'),
    ('yolov8s_1280', 1280, ''),
    ('yolov8s_tiled_640', 640, '--sahi --slice-size 640 --overlap 0.2'),
]:
    pt = find_pt(exp)
    tag = f'{exp}_sahi' if '--sahi' in extra else exp
    out_json = Path(f'preds/{tag}_val.json')
    if pt and not out_json.exists():
        print(f'\n>>> Xuất dự đoán & chấm pycocotools cho {tag} (weights: {pt})')
        !python scripts/predict_yolo.py --weights "{pt}" --gt data/splits/val.json \
            --images data/yolo/images/val --imgsz {imgsz} {extra} \
            --out {out_json} --eval --name {tag}

results_md = Path('reports/eval/results.md')
if results_md.exists():
    print('\n=== BẢNG TỔNG HỢP KẾT QUẢ ĐÁNH GIÁ (pycocotools) ===\n')
    print(results_md.read_text(encoding='utf-8'))

zip_path = Path('../b_yolo_deliverables.zip')
with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zf:
    for p in Path('preds').glob('*.json'):
        zf.write(p)
    for p in Path('reports/eval').glob('*'):
        zf.write(p)
    for p in Path('runs/detect').rglob('weights/best.pt'):
        zf.write(p)
    for p in Path('runs/detect').rglob('results.csv'):
        zf.write(p)

print(f'\nĐã đóng gói toàn bộ dự đoán, bảng chấm điểm và trọng số best.pt vào: {zip_path.resolve()}')
```
